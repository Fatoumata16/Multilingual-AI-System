```python
from sentence_transformers import SentenceTransformer
from transformers import AutoModel, AutoTokenizer
import torch
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Phrases de test
question = "I ye ni sisan kalan ka don."  # "Je veux apprendre maintenant"
true_answer = "Wo ka di! A t'a kɛ kalan kɔnɔ."  # "Bonne réponse"
false_answer = "N bɛ taa so."  # "Fausse réponse"

# === 1. Test avec BGE-M3 ===
def test_bge_m3():
    print("\n🔍 Testing BGE-M3...")
    model = SentenceTransformer("BAAI/bge-m3", device="cpu")  # Ajoutez device="cuda" si vous avez un GPU

    # Encodage
    question_embed = model.encode(question)
    true_embed = model.encode(true_answer)
    false_embed = model.encode(false_answer)

    # Similarité cosinus
    sim_true = cosine_similarity([question_embed], [true_embed])[0][0]
    sim_false = cosine_similarity([question_embed], [false_embed])[0][0]

    print(f"✔️ Similarité avec la BONNE réponse : {sim_true:.4f}")
    print(f"❌ Similarité avec la FAUSSE réponse : {sim_false:.4f}")

# === 2. Test avec AfroXLM-R (comme il n'est pas dans SentenceTransformers, on utilise une approche manuelle) ===
def test_afroxlm_r():
    print("\n🔍 Testing AfroXLM-R...")
    model_name = "Davlan/afro-xlmr-large"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModel.from_pretrained(model_name)

    def mean_pooling(model_output, attention_mask):
        token_embeddings = model_output.last_hidden_state
        input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
        return torch.sum(token_embeddings * input_mask_expanded, 1) / torch.clamp(input_mask_expanded.sum(1), min=1e-9)

    # Encodage des phrases
    def encode(text):
        inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=128)
        with torch.no_grad():
            outputs = model(**inputs)
        return mean_pooling(outputs, inputs['attention_mask']).squeeze(0).numpy()

    question_embed = encode(question)
    true_embed = encode(true_answer)
    false_embed = encode(false_answer)

    # Similarité cosinus
    sim_true = cosine_similarity([question_embed], [true_embed])[0][0]
    sim_false = cosine_similarity([question_embed], [false_embed])[0][0]

    print(f"✔️ Similarité avec la BONNE réponse : {sim_true:.4f}")
    print(f"❌ Similarité avec la FAUSSE réponse : {sim_false:.4f}")

# Lancement des tests
test_bge_m3()
test_afroxlm_r()
```

```python
# =================== 🔧 IMPORTS ===================
import os
import json
import torch
import random
import zipfile
from datetime import datetime
from sklearn.model_selection import train_test_split
from collections import defaultdict

from google.colab import drive
from sentence_transformers import (
    SentenceTransformer,
    InputExample,
    losses,
    evaluation,
    models,
    util
)
from sentence_transformers.datasets import NoDuplicatesDataLoader
from sentence_transformers.evaluation import SequentialEvaluator
from sentence_transformers.util import cos_sim
from torch.utils.data.sampler import BatchSampler, RandomSampler
from transformers import get_linear_schedule_with_warmup
from torch.utils.data import DataLoader
# =================== 📂 CONFIGURATION ===================

MODEL_NAME = "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
BATCH_SIZE = 32
EPOCHS_PHASE1 = 3  # Nombre d'époques pour la phase monolingue
EPOCHS_PHASE2 = 3  # Nombre d'époques pour la phase cross-lingue
LR_PHASE1 = 2e-5
LR_PHASE2 = 5e-6
WARMUP_STEPS = 100
EARLY_STOPPING_PATIENCE = 2
EARLY_STOPPING_MIN_DELTA = 0.01

# Chemins des fichiers
TRIPLET_ZIP_PATH = "/content/drive/MyDrive/culture_donnees_sur_drive_sans_variance.zip"
EXTRACT_DIR = "/content/triplet_data"
SAVE_DIR = "./fine_tuned_model_bambara"
LANGUES_ACCEPTEES = {'français', 'bambara'}

from huggingface_hub import notebook_login
notebook_login()
# =================== 🔗 MONTAGE GOOGLE DRIVE ===================
print("🚀 Montage de Google Drive...")
drive.mount('/content/drive')

# =================== 🎮 CONFIGURATION GPU ===================
print("\n🔍 Vérification de la disponibilité du GPU...")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device utilisé: {device}")
if device.type == "cuda":
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"Capabilité CUDA: {torch.cuda.get_device_capability(0)}")
    print(f"Mémoire disponible: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.2f} Go")


# =================== 🔧 CONFIGURATION INITIALE ===================
import os
# Désactive complètement Weights & Biases
os.environ["WANDB_DISABLED"] = "true"

# =================== 📦 EXTRACTION DU ZIP ===================
print("📁 Extraction des données du fichier zip...")
os.makedirs(EXTRACT_DIR, exist_ok=True)
with zipfile.ZipFile(TRIPLET_ZIP_PATH, 'r') as zip_ref:
    zip_ref.extractall(EXTRACT_DIR)

# Mettre à jour le chemin vers le fichier JSON une fois extrait
TRIPLETS_PATH = os.path.join(EXTRACT_DIR, 'culture_donnees_sur_drive_sans_variance.json')

# =================== 📚 CHARGEMENT ET NETTOYAGE DES TRIPLETS ===================
def load_and_clean_triplets(filepath):
    """
    Charge les triplets depuis un fichier JSON, nettoie le texte,
    supprime les triplets dupliqués exacts et valide la structure.
    """
    with open(filepath, 'r', encoding='utf-8') as f:
        raw_data = json.load(f)

    cleaned = []
    seen = set()
    id_counts = defaultdict(int)

    required_keys = {'anchor', 'positive', 'negative', 'langue'}
    optional_keys = {'positive_langue', 'id', 'niveau', 'matiere', 'negative_type'}

    for idx, t in enumerate(raw_data):
        # Validation des champs obligatoires
        if not all(k in t for k in required_keys):
            print(f"⚠️ Triplet incomplet ignoré (ligne {idx}): {t.get('id', 'sans ID')}")
            continue

        # Nettoyage des textes
        anchor = t['anchor'].strip()
        positive = t['positive'].strip()
        negative = t['negative'].strip()
        langue = t['langue'].lower()
        positive_langue = t.get("positive_langue", langue).lower()

        # Validation des langues
        if langue not in LANGUES_ACCEPTEES or positive_langue not in LANGUES_ACCEPTEES:
            print(f"🚫 Langue non acceptée - Anchor: {langue}, Positive: {positive_langue}")
            continue

        # Génération d'un ID unique si non fourni
        if 'id' not in t:
            base_id = f"{t.get('matiere', 'GEN')}_{langue[:2]}_{idx}"
            t['id'] = base_id

        # Comptage des IDs pour détecter les doublons
        id_counts[t['id']] += 1
        if id_counts[t['id']] > 1:
            t['id'] = f"{t['id']}_{id_counts[t['id']]}"

        # Création d'une signature unique pour détecter les doublons
        content_key = (anchor[:100], positive[:100], negative[:100], langue, positive_langue)

        if content_key not in seen:
            seen.add(content_key)
            cleaned.append({
                'id': t['id'],
                'anchor': anchor,
                'positive': positive,
                'negative': negative,
                'langue': langue,
                'positive_langue': positive_langue,
                'niveau': t.get('niveau'),
                'matiere': t.get('matiere'),
                'negative_type': t.get('negative_type')
            })
        else:
            print(f"🚫 Doublon détecté - ID: {t['id']}")

    print(f"\n✔️ {len(cleaned)} triplets uniques et valides chargés.")
    print(f"📊 Répartition par langue:")
    for lang in LANGUES_ACCEPTEES:
        count = sum(1 for t in cleaned if t['langue'] == lang)
        print(f"  - {lang.capitalize()}: {count} triplets")

    return cleaned

# =================== 🧪 DATASET SPLIT (MONOLINGUE / CROSS-LANGUE) ===================
def split_data(triplets):
    """
    Sépare les triplets en ensembles monolingues (Phase 1) et cross-lingues (Phase 2).
    """
    phase1 = []  # Triplés où anchor_langue == positive_langue
    phase2 = []  # Triplés où anchor_langue != positive_langue

    for t in triplets:
        if t["langue"] == t["positive_langue"]:
            phase1.append(t)
        else:
            phase2.append(t)

    print(f"\n📊 Division des données:")
    print(f"  - Phase 1 (monolingue): {len(phase1)} triplets")
    print(f"  - Phase 2 (cross-lingue): {len(phase2)} triplets")

    # Statistiques supplémentaires
    if phase1:
        print("\n🔵 Phase 1 - Répartition:")
        for lang in LANGUES_ACCEPTEES:
            count = sum(1 for t in phase1 if t['langue'] == lang)
            print(f"  - {lang.capitalize()}: {count} triplets")

    if phase2:
        print("\n🟣 Phase 2 - Paires linguistiques:")
        cross_pairs = set((t['langue'], t['positive_langue']) for t in phase2)
        for pair in cross_pairs:
            count = sum(1 for t in phase2 if t['langue'] == pair[0] and t['positive_langue'] == pair[1])
            print(f"  - {pair[0]} → {pair[1]}: {count} triplets")

    return phase1, phase2

# =================== 📝 CONVERSION EN INPUTEXAMPLES ===================
def to_input_examples(triplets_list):
    """Convertit une liste de dictionnaires triplets en objets InputExample."""
    examples = []
    for t in triplets_list:
        # Validation de la longueur des textes
        if len(t["anchor"]) < 5 or len(t["positive"]) < 5 or len(t["negative"]) < 5:
            print(f"⚠️ Texte trop court ignoré - ID: {t['id']}")
            continue
        examples.append(InputExample(
            texts=[t["anchor"], t["positive"], t["negative"]],
            label=1.0  # Label fixe pour les triplets
        ))
    return examples

# =================== ⚙️ LOSS FUNCTIONS CUSTOMISÉES ===================
class DynamicTripletLoss(losses.TripletLoss):
    """
    Triplet Loss avec une marge dynamique basée sur la difficulté des exemples.
    """
    def __init__(self, model, **kwargs):
        super().__init__(model, **kwargs)
        self.min_margin = 0.2
        self.max_margin = 0.8
        self.margin_history = []

    def forward(self, sentence_features, labels):
        # Calcul des embeddings
        embeddings = [self.model(sf)['sentence_embedding'] for sf in sentence_features]
        anchor, positive, negative = embeddings

        # Calcul des similarités
        sim_pos = cos_sim(anchor, positive)
        sim_neg = cos_sim(anchor, negative)

        # Ajustement dynamique de la marge
        current_diff = (sim_neg - sim_pos).mean().item()
        dynamic_margin = min(max(current_diff, self.min_margin), self.max_margin)
        self.triplet_margin = dynamic_margin
        self.margin_history.append(dynamic_margin)

        # Calcul de la perte avec la marge dynamique
        loss = torch.relu(sim_pos - sim_neg + dynamic_margin)
        return loss.mean()

class AlignedCombinedLoss(torch.nn.Module):
    """
    Combine TripletLoss, MultipleNegativesRankingLoss et un terme d'alignement cross-lingue.
    """
    def __init__(self, model, triplet_weight=1.0, mnr_weight=0.5, align_weight=0.3):
        super().__init__()
        self.model = model
        self.triplet = DynamicTripletLoss(model=model)
        self.mnr = losses.MultipleNegativesRankingLoss(model=model)
        self.triplet_weight = triplet_weight
        self.mnr_weight = mnr_weight
        self.align_weight = align_weight

    def forward(self, sentence_features, labels):
        # Calcul des embeddings
        embeddings = [self.model(sf)['sentence_embedding'] for sf in sentence_features]
        anchor, positive, negative = embeddings

        # Calcul des différentes pertes
        triplet_loss = self.triplet(sentence_features, labels)

        # MNR Loss nécessite une reformulation des données
        mnr_features = [{'sentence_embedding': anchor}, {'sentence_embedding': positive}]
        mnr_loss = self.mnr(mnr_features, labels)

        # Perte d'alignement cross-lingue
        align_loss = 1 - cos_sim(anchor, positive).mean()

        # Combinaison pondérée
        total_loss = (self.triplet_weight * triplet_loss +
                     self.mnr_weight * mnr_loss +
                     self.align_weight * align_loss)

        # Monitoring (pour debug)
        if random.random() < 0.01:  # Log 1% des batches
            print(f"\n🔍 Monitoring des pertes:")
            print(f"  - Triplet: {triplet_loss.item():.4f} (marge: {self.triplet.triplet_margin:.3f})")
            print(f"  - MNR: {mnr_loss.item():.4f}")
            print(f"  - Alignement: {align_loss.item():.4f}")
            print(f"  → Total: {total_loss.item():.4f}")

        return total_loss

# =================== 📊 ÉVALUATEURS ROBUSTES ===================
def build_evaluation_sets():
    """Construit des ensembles de test pour l'évaluation."""
    # Exemples de requêtes et corpus pour chaque langue
    queries = {
        "fr": [
            "Quelle est la boisson chaude emblématique de l'hospitalité malienne ?",
            "Comment s'appelle le vêtement traditionnel porté au Mali ?",
            "Quel est le rôle du cerveau dans la motricité ?",
            "Définissez une tempête de sable."
        ],
        "bm": [
            "Mali faso minfɛn ye jumɛn ye, min bɛ sɔrɔ seli bɛɛ la?",
            "Mali laadala fini min ka bon ani min ɲɛw ye bulama ye, an b'o wele cogo di?",
            "Hakili bɛ an ka lamaga-lamagaliw kɔlɔsi cogo di?",
            "Cɛɲi ye mun ye?"
        ]
    }

    corpus = {
        "fr": [
            "Le thé vert, appelé 'ataya', est la boisson la plus consommée au Mali et symbolise l'hospitalité.",
            "Le boubou est le vêtement traditionnel malien, porté en toutes occasions par les hommes et les femmes.",
            "Le cerveau contrôle les mouvements volontaires et involontaires via le cortex moteur et le cervelet.",
            "Une tempête de sable est un phénomène météorologique avec des vents forts transportant du sable."
        ],
        "bm": [
            "Te wuluwulu, n'a bɛ fɔ a ma ko 'ataya', ye Mali ka minfɛn ye min bɛ ji caman ye, ani a bɛ jira bonya ni hakili la.",
            "Boubou ye Mali laadala fini ye, min bɛ fɛnw caya ani min bɛ fɛnw caya.",
            "Sugunɛ, ani n'a fɔra ka ɲɛ, farikolo-ɲɛnajɛ-yɔrɔ ni kunsɛmɛ, jɔyɔrɔ ka bon kosɛbɛ ŋaniyajira ni ŋaniyabaliya lamaga-lamagaliw kɔlɔsili la.",
            "Cɛɲi ye waati ye min na fiɲɛba bɛ cɛncɛn ni buguri camanba lawuli ka bɔ duguma."
        ]
    }

    # Documents pertinents pour chaque requête (dans les deux langues)
    relevant_docs = {
        # Requêtes françaises
        queries["fr"][0]: [corpus["fr"][0], corpus["bm"][0]],
        queries["fr"][1]: [corpus["fr"][1], corpus["bm"][1]],
        queries["fr"][2]: [corpus["fr"][2], corpus["bm"][2]],
        queries["fr"][3]: [corpus["fr"][3], corpus["bm"][3]],
        # Requêtes bambara
        queries["bm"][0]: [corpus["fr"][0], corpus["bm"][0]],
        queries["bm"][1]: [corpus["fr"][1], corpus["bm"][1]],
        queries["bm"][2]: [corpus["fr"][2], corpus["bm"][2]],
        queries["bm"][3]: [corpus["fr"][3], corpus["bm"][3]]
    }

    # Création des dictionnaires nécessaires pour les évaluateurs
    corpus_dict = {f"fr_{i}": text for i, text in enumerate(corpus["fr"])}
    corpus_dict.update({f"bm_{i}": text for i, text in enumerate(corpus["bm"])})

    relevant_docs_mapped = {}
    for q_text, docs in relevant_docs.items():
        relevant_docs_mapped[q_text] = []
        for doc in docs:
            for doc_id, text in corpus_dict.items():
                if text == doc:
                    relevant_docs_mapped[q_text].append(doc_id)
                    break

    return queries, corpus_dict, relevant_docs_mapped

def build_robust_evaluators():
    """Construit plusieurs évaluateurs pour un suivi complet."""
    queries, corpus, relevant_docs = build_evaluation_sets()

    # Évaluateur monolingue français
    eval_fr = evaluation.InformationRetrievalEvaluator(
        queries["fr"],
        {k: v for k, v in corpus.items() if k.startswith('fr_')},
        {k: [d for d in v if d.startswith('fr_')]
         for k, v in relevant_docs.items() if k in queries["fr"]},
        name="French Monolingual",
        show_progress_bar=False
    )

    # Évaluateur monolingue bambara
    eval_bm = evaluation.InformationRetrievalEvaluator(
        queries["bm"],
        {k: v for k, v in corpus.items() if k.startswith('bm_')},
        {k: [d for d in v if d.startswith('bm_')]
         for k, v in relevant_docs.items() if k in queries["bm"]},
        name="Bambara Monolingual",
        show_progress_bar=False
    )

    # Évaluateur cross-lingue complet
    eval_cross = evaluation.InformationRetrievalEvaluator(
        queries["fr"] + queries["bm"],
        corpus,
        relevant_docs,
        name="Cross-lingual",
        show_progress_bar=False
    )

    # Évaluateur séquentiel combinant les trois
    return SequentialEvaluator([eval_fr, eval_bm, eval_cross])

# =================== 🏋️ DATALOADERS ÉQUILIBRÉS ===================
class BalancedTripletSampler:
    """Garantit un équilibre des langues dans chaque batch."""
    def __init__(self, triplets, batch_size):
        self.triplets = triplets
        self.batch_size = batch_size
        self.lang_indices = {
            lang: [i for i, t in enumerate(triplets) if t['langue'] == lang]
            for lang in LANGUES_ACCEPTEES
        }

    def __iter__(self):
        # Mélanger les indices pour chaque langue
        for lang in self.lang_indices:
            random.shuffle(self.lang_indices[lang])

        # Calculer le nombre de batches complets possibles
        min_samples = min(len(indices) for indices in self.lang_indices.values())
        num_batches = min_samples // (self.batch_size // len(LANGUES_ACCEPTEES))

        # Générer des batches équilibrés
        for i in range(num_batches):
            batch = []
            for lang in self.lang_indices:
                start = i * (self.batch_size // len(LANGUES_ACCEPTEES))
                end = start + (self.batch_size // len(LANGUES_ACCEPTEES))
                batch.extend(self.lang_indices[lang][start:end])

            random.shuffle(batch)
            yield batch

    def __len__(self):
        min_samples = min(len(indices) for indices in self.lang_indices.values())
        return min_samples // (self.batch_size // len(LANGUES_ACCEPTEES))

def create_balanced_dataloader(triplets, examples, batch_size, phase_name):
    """Crée un DataLoader équilibré selon la phase d'entraînement."""
    if phase_name == "monolingual":
        # Pour la phase monolingue, nous voulons des batches mixtes français/bambara
        sampler = BalancedTripletSampler(triplets, batch_size)
        return DataLoader(
            examples,
            batch_sampler=BatchSampler(sampler, batch_size=batch_size, drop_last=False),
            shuffle=False
        )
    else:
        # Pour la phase cross-lingue, nous utilisons NoDuplicatesDataLoader standard
        return NoDuplicatesDataLoader(examples, batch_size=batch_size)

# =================== 🏁 FONCTION D'ENTRAÎNEMENT ===================
def fit_phase(model, dataloader, triplets, epochs, lr, phase_name, evaluator):
    """
    Fonction complète pour entraîner le modèle sur une phase donnée.
    """
    print(f"\n🏁 Début de la phase {phase_name}...")
    print(f"  - Nombre de batches par epoch: {len(dataloader)}")
    print(f"  - Taille des batches: {dataloader.batch_size}")
    print(f"  - Nombre d'époques: {epochs}")
    print(f"  - Taux d'apprentissage: {lr:.1e}")

    # Optimizer et scheduler
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=WARMUP_STEPS,
        num_training_steps=epochs * len(dataloader)
    )

    # Configuration de l'early stopping
    early_stopping = evaluation.EarlyStopping(
        checkpoint_path=os.path.join(SAVE_DIR, f"best_{phase_name}"),
        patience=EARLY_STOPPING_PATIENCE,
        min_delta=EARLY_STOPPING_MIN_DELTA,
        max_epochs=epochs
    )

    # Configuration de la perte
    if phase_name == "monolingual":
        loss_model = AlignedCombinedLoss(model, triplet_weight=1.0, mnr_weight=0.5, align_weight=0.0)
    else:
        loss_model = AlignedCombinedLoss(model, triplet_weight=1.0, mnr_weight=0.3, align_weight=0.3)

    # Entraînement
    model.fit(
        train_objectives=[(dataloader, loss_model)],
        epochs=epochs,
        warmup_steps=WARMUP_STEPS,
        optimizer=optimizer,
        scheduler=scheduler,
        evaluator=evaluator,
        evaluation_steps=max(50, len(dataloader)//5),  # Évaluer 5 fois par epoch
        output_path=os.path.join(SAVE_DIR, phase_name),
        save_best_model=True,
        use_amp=True,
        show_progress_bar=True,
        callback=[early_stopping]
    )

    print(f"✅ Phase {phase_name} terminée. Meilleur modèle sauvegardé.")

# =================== 🧪 TESTS FINAUX ===================
def run_final_tests(model):
    """Exécute une batterie de tests finaux sur le modèle entraîné."""
    print("\n🔬 Lancement des tests finaux...")

    # Test d'alignement cross-lingue
    print("\n🔍 Test d'alignement cross-lingue:")
    test_pairs = [
        ("Le thé malien est très populaire.", "Te Mali ka ka bon kosɛbɛ."),
        ("Le boubou est un vêtement traditionnel élégant.", "Boubou ye fini laadala ye min ka ɲi."),
        ("La capitale du Mali est Bamako.", "Mali faaba ye Bamako ye."),
        ("Comment dit-on 'bonjour' en bambara ?", "I ni ce kɛlɛbɛ la min bɛ fɔ bambara kan na?")
    ]

    for fr, bm in test_pairs:
        fr_emb = model.encode(fr, convert_to_tensor=True)
        bm_emb = model.encode(bm, convert_to_tensor=True)
        sim = cos_sim(fr_emb, bm_emb).item()
        print(f"  {fr[:30]}... ↔ {bm[:30]}...: {sim:.4f}")

    # Test de similarité monolingue
    print("\n🔍 Test de similarité monolingue:")
    fr_pairs = [
        ("Le thé malien est délicieux.", "L'ataya est une boisson populaire au Mali."),
        ("Le boubou est un vêtement ample.", "Les Maliens portent des vêtements traditionnels.")
    ]

    for a, b in fr_pairs:
        a_emb = model.encode(a, convert_to_tensor=True)
        b_emb = model.encode(b, convert_to_tensor=True)
        sim = cos_sim(a_emb, b_emb).item()
        print(f"  {a[:30]}... ↔ {b[:30]}...: {sim:.4f}")

    bm_pairs = [
        ("Te Mali ka di.", "Ataya ye minfɛn ye min bɛ Mali la."),
        ("Boubou ye fini dɔ ye.", "Mali denw bɛ fini caman wɛrɛw sɔrɔ.")
    ]

    for a, b in bm_pairs:
        a_emb = model.encode(a, convert_to_tensor=True)
        b_emb = model.encode(b, convert_to_tensor=True)
        sim = cos_sim(a_emb, b_emb).item()
        print(f"  {a[:30]}... ↔ {b[:30]}...: {sim:.4f}")

# =================== 🚀 PIPELINE PRINCIPAL ===================
def main():
    # Initialisation
    os.makedirs(SAVE_DIR, exist_ok=True)
    print("\n" + "="*50)
    print("🚀 INITIALISATION DU FINE-TUNING MULTILINGUE")
    print("="*50)

    # 1. Chargement et préparation des données
    print("\n📦 CHARGEMENT ET PRÉPARATION DES DONNÉES")
    triplets = load_and_clean_triplets(TRIPLETS_PATH)
    phase1_triplets, phase2_triplets = split_data(triplets)

    # 2. Conversion en InputExamples
    phase1_examples = to_input_examples(phase1_triplets)
    phase2_examples = to_input_examples(phase2_triplets)

    # 3. Création des DataLoaders équilibrés
    print("\n⚖️ CRÉATION DES DATALOADERS ÉQUILIBRÉS")
    dataloader1 = create_balanced_dataloader(phase1_triplets, phase1_examples, BATCH_SIZE, "monolingual")
    dataloader2 = create_balanced_dataloader(phase2_triplets, phase2_examples, BATCH_SIZE, "crosslingual")

    # 4. Initialisation du modèle
    print("\n🔄 CHARGEMENT DU MODÈLE DE BASE")
    word_embed = models.Transformer(MODEL_NAME)
    pooling = models.Pooling(word_embed.get_word_embedding_dimension(), 'cls')
    model = SentenceTransformer(modules=[word_embed, pooling])

    # Déplacer le modèle sur le GPU si disponible
    model.to(device)
    print(f"Modèle chargé: {MODEL_NAME}")
    print(f"Device utilisé: {next(model.parameters()).device}")
    print(f"Dimension des embeddings: {model.get_sentence_embedding_dimension()}")

    # 5. Préparation des évaluateurs
    evaluator = build_robust_evaluators()

    # 6. Phase 1 - Entraînement monolingue
    print("\n" + "="*50)
    print("🔵 PHASE 1 - ENTRAÎNEMENT MONOLINGUE")
    print("="*50)
    fit_phase(model, dataloader1, phase1_triplets, EPOCHS_PHASE1, LR_PHASE1, "phase1_monolingual", evaluator)

    # 7. Phase 2 - Entraînement cross-lingue
    print("\n" + "="*50)
    print("🟣 PHASE 2 - ENTRAÎNEMENT CROSS-LINGUE")
    print("="*50)
    fit_phase(model, dataloader2, phase2_triplets, EPOCHS_PHASE2, LR_PHASE2, "phase2_crosslingual", evaluator)

    # 8. Tests finaux
    print("\n" + "="*50)
    print("🧪 TESTS FINAUX")
    print("="*50)
    run_final_tests(model)

    # 9. Sauvegarde finale
    final_path = os.path.join(SAVE_DIR, "final_model")
    model.save(final_path)
    print(f"\n✅ ENTRAÎNEMENT TERMINÉ. Modèle final sauvegardé dans: {final_path}")

if __name__ == "__main__":
    main()
```

```python
import os
import json
import zipfile
import random
from datasets import Dataset
from transformers import EarlyStoppingCallback
from sentence_transformers import SentenceTransformer, losses, util, evaluation
from sentence_transformers.trainer import SentenceTransformerTrainer
from sentence_transformers.training_args import SentenceTransformerTrainingArguments

# ========== CONFIG ==========
MODEL_NAME = "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
TRIPLETS_ZIP_PATH = "/content/drive/MyDrive/culture_donnees_sur_drive_sans_variance.zip"
SAVE_DIR = "./fine_tuned_model_bambara"
os.makedirs(SAVE_DIR, exist_ok=True)

BATCH_SIZE = 16
EPOCHS_PHASE1 = 10  # Augmenté pour permettre un meilleur apprentissage
EPOCHS_PHASE2 = 8
LR_PHASE1 = 2e-5    # Learning rate légèrement augmenté
LR_PHASE2 = 1e-5
MIN_DELTA = 0.0001  # Plus sensible pour détecter les améliorations

# ========== DATA AUGMENTATION ==========
def augment_triplet(triplet):
    """Simple data augmentation for text triplets"""
    augmented = triplet.copy()

    # Random word deletion (10% chance per word)
    for field in ['anchor', 'positive', 'negative']:
        words = triplet[field].split()
        if len(words) > 3 and random.random() < 0.3:
            del_idx = random.randint(0, len(words)-1)
            words.pop(del_idx)
            augmented[field] = ' '.join(words)

    # Synonym replacement (simplified example)
    replacements = {
        'bonjour': ['salam', 'aw ni sogoma'],
        'merci': ['a ni ce', 'i ni ce']
    }

    for field in ['anchor', 'positive', 'negative']:
        for word, syns in replacements.items():
            if word in triplet[field]:
                augmented[field] = augmented[field].replace(word, random.choice(syns))

    return augmented

# ========== ENHANCED EVALUATOR ==========
def build_robust_evaluator(dataset, num_samples=200):
    """Build evaluator with proper train/test split"""
    # Sample evaluation data
    eval_samples = dataset.shuffle().select(range(min(num_samples, len(dataset))))

    queries = {}
    corpus = {}
    relevant_docs = {}

    for idx, sample in enumerate(eval_samples):
        queries[idx] = sample['anchor']
        corpus[2*idx] = sample['positive']
        corpus[2*idx+1] = sample['negative']
        relevant_docs[idx] = [2*idx]  # Only positive is relevant

    return evaluation.InformationRetrievalEvaluator(
        queries, corpus, relevant_docs,
        show_progress_bar=True,
        corpus_chunk_size=512,
        precision_recall_at_k=[3, 5, 10],
        name='bambara_eval'
    )

# ========== IMPROVED DATA LOADING ==========
def load_and_prepare_data():
    with zipfile.ZipFile(TRIPLETS_ZIP_PATH, 'r') as z:
        with z.open(z.namelist()[0]) as f:
            data = json.load(f)

    # Apply simple data augmentation
    augmented_data = []
    for item in data:
        augmented_data.append(item)
        # Add 1 augmented version for each original triplet (50% increase)
        if random.random() < 0.5:
            augmented_data.append(augment_triplet(item))

    # Split into phases with 80/20 train/val split
    random.shuffle(augmented_data)
    split_idx = int(0.8 * len(augmented_data))

    phase1_data, phase2_data = [], []
    for item in augmented_data[:split_idx]:  # Training data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            phase1_data.append(record)
        else:
            phase2_data.append(record)

    # Create validation sets
    val_phase1, val_phase2 = [], []
    for item in augmented_data[split_idx:]:  # Validation data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            val_phase1.append(record)
        else:
            val_phase2.append(record)

    return (
        Dataset.from_list(phase1_data),
        Dataset.from_list(val_phase1),
        Dataset.from_list(phase2_data),
        Dataset.from_list(val_phase2)
    )

# ========== ENHANCED TRAINING PHASE ==========
def train_phase(model, train_data, val_data, epochs, lr, phase_name):
    # Use multiple negative ranking loss for better convergence
    loss = losses.MultipleNegativesRankingLoss(model)

    # Create robust evaluator
    evaluator = build_robust_evaluator(val_data)

    # Enhanced training arguments
    training_args = SentenceTransformerTrainingArguments(
        output_dir=os.path.join(SAVE_DIR, phase_name),
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE*2,
        num_train_epochs=epochs,
        learning_rate=lr,
        weight_decay=0.01,  # Added regularization
        warmup_ratio=0.1,   # Better learning rate scheduling
        # evaluation_strategy="steps",
        eval_strategy="steps",
        logging_strategy="steps",
        eval_steps=100,
        save_steps=200,
        save_total_limit=2,
        load_best_model_at_end=True,
        metric_for_best_model="eval_bambara_eval_cosine_ndcg@10",
        greater_is_better=True,
        logging_dir=os.path.join(SAVE_DIR, "logs"),
        logging_steps=50,
        report_to=[],
        fp16=True  # Enable mixed precision training
    )

    trainer = SentenceTransformerTrainer(
        model=model,
        args=training_args,
        train_dataset=train_data,
        eval_dataset=val_data,
        loss=loss,
        evaluator=evaluator,
        callbacks=[
            EarlyStoppingCallback(
                early_stopping_patience=3,
                early_stopping_threshold=MIN_DELTA
            )
        ],
    )

    print(f"\n⭐ Starting {phase_name} training with {len(train_data)} samples")
    trainer.train()

    # Final evaluation
    print("\n📊 Final evaluation results:")
    evaluator(model)

    model.save(os.path.join(SAVE_DIR, f"{phase_name}_final"))
    return model

# ========== MAIN ==========
def main():
    # Load model with better tokenizer handling
    model = SentenceTransformer(MODEL_NAME)
    print("Model loaded with {} parameters".format(sum(p.numel() for p in model.parameters())))

    # Load and split data
    phase1_train, phase1_val, phase2_train, phase2_val = load_and_prepare_data()
    print(f"\nData loaded: {len(phase1_train)} phase1 train, {len(phase1_val)} phase1 val, "
          f"{len(phase2_train)} phase2 train, {len(phase2_val)} phase2 val samples")

    # Phase 1: Monolingual fine-tuning
    print("\n🔵 Phase 1: Monolingual Fine-tuning")
    model = train_phase(
        model, phase1_train, phase1_val,
        EPOCHS_PHASE1, LR_PHASE1, "phase1"
    )

    # Phase 2: Cross-lingual adaptation
    print("\n🟣 Phase 2: Cross-lingual Adaptation")
    model = train_phase(
        model, phase2_train, phase2_val,
        EPOCHS_PHASE2, LR_PHASE2, "phase2"
    )

    # Final save with better metadata
    model.save(os.path.join(SAVE_DIR, "final_model_bambara"))
    print("\n✅ Training completed successfully!")

    # Test some examples
    test_queries = [
        ("Bonjour comment ça va?", "I ni sogoma, i ka kɛnɛ wa?"),
        ("Quelle est la capitale du Mali?", "Mali faaba ye mun ye?")
    ]
    for q1, q2 in test_queries:
        emb1 = model.encode(q1)
        emb2 = model.encode(q2)
        sim = util.cos_sim(emb1, emb2).item()
        print(f"Similarity between '{q1}' and '{q2}': {sim:.4f}")

if __name__ == "__main__":
    main()
```

```python
# ici nous allons ajouter les mots les plus frequents de notre corpus bambara dans le tokenizer et refaire un entrainement pour voir si nous avons de meilleurs resultats
import os
import json
import zipfile
import random
from datasets import Dataset
from transformers import EarlyStoppingCallback
from sentence_transformers import SentenceTransformer, losses, util, evaluation
from sentence_transformers.trainer import SentenceTransformerTrainer
from sentence_transformers.training_args import SentenceTransformerTrainingArguments
from google.colab import drive

# ========== CONFIG ==========
MODEL_NAME = "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
TRIPLETS_ZIP_PATH = "/content/drive/MyDrive/culture_donnees_sur_drive_sans_variance.zip"
SAVE_DIR = "/content/drive/MyDrive/fine_tuned_model_bambara_fusion3"
os.makedirs(SAVE_DIR, exist_ok=True)

# Liste des tokens bambara à ajouter
BAMBARA_TOKENS = [
    "_tiɲɛ",
    "_bɔli",
    "_daɲɛw",
    "_jatigɛ",
    "_jɛkulu",
    "_kɛlɛ",
    "_kɛrɛnkɛrɛnnenya",
    "_yɔrɔ",
    "_baarakɛcogo",
    "_baarakɛlaw",
    "_baarakɛnafolo",
    "_balofɛn",
    "_balofɛnw",
    "_banakisɛ",
    "_banakisɛw",
    "_bolifɛnw",
    "_bololabaarakɛlaw",
    "_bɔ",
    "_bɔgɔ",
    "_bɔgɔdaga",
    "_bɔlen",
    "_bɔnsɔn",
    "_bɔnsɔnw",
    "_bɔnɛni",
    "_bɔra",
    "_bɔyɔrɔ",
    "_bɛ",
    "_bɛn",
    "_bɛna",
    "_bɛnbaliya",
    "_bɛnbaliyakow",
    "_bɛnbaw",
    "_bɛnkan",
    "_bɛnnen",
    "_bɛɛ",
    "_ciyɛn",
    "_cɛ",
    "_cɛfarinw",
    "_cɛma",
    "_cɛmancɛ",
    "_cɛsira",
    "_cɛsiri",
    "_cɛw",
    "_cɛɲi",
    "_dabɔ",
    "_dabɔli",
    "_dabɔra",
    "_dalajɛli",
    "_daminɛ",
    "_daminɛna",
    "_dancɛw",
    "_danfɛnw",
    "_dantigɛli",
    "_danyɔrɔ",
    "_daɲɛ",
    "_daɲɛgafe",
    "_denmisɛnw",
    "_diinɛ",
    "_diinɛko",
    "_diinɛw",
    "_diɲɛ",
    "_dumunikɛcogo",
    "_dumunikɛyɔrɔ",
    "_dɔ",
    "_dɔgɔya",
    "_dɔgɔyali",
    "_dɔn",
    "_dɔnbaliya",
    "_dɔnkili",
    "_dɔnkilida",
    "_dɔnkilidala",
    "_dɔnkiliw",
    "_dɔnko",
    "_dɔnkotigi",
    "_dɔnni",
    "_dɔnniya",
    "_dɔnniyaw",
    "_dɔnw",
    "_dɔrɔn",
    "_dɔw",
    "_dɔni",
    "_dɛmɛ",
    "_dɛsɛ",
    "_erɔpu",
    "_erɔpukaw",
    "_farikolojɔli",
    "_farikoloɲɛnajɛ",
    "_fasodenɲumanya",
    "_fiɲɛ",
    "_fiɲɛbɔlan",
    "_funankɛninw",
    "_furakɛli",
    "_fɔcogo",
    "_fɔli",
    "_fɔlifɛn",
    "_fɔlifɛnw",
    "_fɔlisen",
    "_fɔlɔ",
    "_fɛ",
    "_fɛn",
    "_fɛnba",
    "_fɛnkɔrɔw",
    "_fɛnw",
    "_fɛnɲɛnama",
    "_fɛnɲɛnamafagalan",
    "_fɛnɲɛnamafagalanw",
    "_fɛnɲɛnamaw",
    "_fɛɛrɛ",
    "_fɛɛrɛw",
    "_gɛlɛya",
    "_gɛlɛyabaw",
    "_gɛlɛyaw",
    "_hakilijagabɔ",
    "_hakɛ",
    "_hɔrɔnya",
    "_hɛrɛ",
    "_jagɛlɛya",
    "_jatebɔ",
    "_jateminɛ",
    "_jinɛw",
    "_jisigiyɔrɔw",
    "_jogoɲumanya",
    "_jumɛn",
    "_jɔ",
    "_jɔcogo",
    "_jɔlen",
    "_jɔli",
    "_jɔn",
    "_jɔnfeere",
    "_jɔnjɔn",
    "_jɔnjɔnw",
    "_jɔnw",
    "_jɔnya",
    "_jɔyɔrɔ",
    "_jɔyɔrɔw",
    "_jɛgɛ",
    "_jɛkafɔ",
    "_jɛkuluw",
    "_jɛlen",
    "_jɛman",
    "_jɛmukan",
    "_jɛnsɛn",
    "_jɛnsɛnni",
    "_jɛɲɔgɔnya",
    "_jɛɲɔgɔnyaw",
    "_kemikɛlifɛnw",
    "_kiritigɛ",
    "_kiritigɛlaw",
    "_kosɔn.",
    "_kosɛbɛ",
    "_kulɛri",
    "_kumaɲɔgɔnya",
    "_kunbɛnni",
    "_kɔ",
    "_kɔfɛ",
    "_kɔgɔji",
    "_kɔgɔjida",
    "_kɔkan",
    "_kɔkanna",
    "_kɔlɔlɔ",
    "_kɔlɔlɔw",
    "_kɔlɔsili",
    "_kɔnɔ",
    "_kɔnɔfɛnw",
    "_kɔnɔkow",
    "_kɔnɔna",
    "_kɔnɔw",
    "_kɔrɔ",
    "_kɔrɔfɔli",
    "_kɔrɔlen",
    "_kɔrɔnyanfan",
    "_kɔrɔw",
    "_kɔɲɔw",
    "_kɛ",
    "_kɛcogo",
    "_kɛlen",
    "_kɛlenw",
    "_kɛli",
    "_kɛlɛba",
    "_kɛlɛli",
    "_kɛlɛw",
    "_kɛmɛ",
    "_kɛnɛ",
    "_kɛnɛma",
    "_kɛnɛman",
    "_kɛnɛya",
    "_kɛnɛyako",
    "_kɛrɛnkɛrɛnnen",
    "_kɛrɛnkɛrɛnnenw",
    "_kɛwalew",
    "_laadalatɔn",
    "_laadalatɔnw",
    "_labɛn",
    "_labɛncogo",
    "_labɛnnen",
    "_labɛnni",
    "_lajɛ",
    "_lajɛba",
    "_lajɛlen",
    "_lakɔli",
    "_lamɛnbagaw",
    "_lamɛnni",
    "_lamɛnnikɛlaw",
    "_latigɛw",
    "_laɲini",
    "_maanafɔla",
    "_maanafɔlaw",
    "_masakɛ",
    "_maskɛw",
    "_minfɛn",
    "_minɛ",
    "_minɛni",
    "_minɛnw",
    "_misɛnninw",
    "_muɲuli",
    "_mɔgɔ",
    "_mɔgɔba",
    "_mɔgɔkɔrɔbaw",
    "_mɔgɔw",
    "_mɔni",
    "_nafabɔli",
    "_nakɔfɛnw",
    "_nbɛdaw",
    "_nisɔndiya",
    "_nɔba",
    "_nɔfɛkow",
    "_nɔgɔ",
    "_nɔgɔlenw",
    "_nɔgɔli",
    "_nɔgɔman",
    "_nɔgɔmanw",
    "_nɔgɔn",
    "_nɔgɔw",
    "_nɔgɔya",
    "_nɛgɛ",
    "_nɛgɛso",
    "_nɛgɛw",
    "_nɛnɛ",
    "_politikitɔnw",
    "_revolisiyɔn",
    "_sanfɛla",
    "_senfɛ",
    "_sigidamɔgɔw",
    "_sigiyɔrɔ",
    "_silamɛ",
    "_silamɛw",
    "_silamɛya",
    "_siniɲɛsigi",
    "_siratigɛ",
    "_siɲɛ",
    "_sunɔgɔ",
    "_sɔgɔsɔgɔninjɛ",
    "_sɔn",
    "_sɔngɔ",
    "_sɔrɔ",
    "_sɔrɔcogo",
    "_sɔrɔko",
    "_sɔrɔli",
    "_sɔrɔta",
    "_sɔrɔyɔrɔ",
    "_sɔsɔli",
    "_sɛbɛn",
    "_sɛbɛnni",
    "_sɛbɛnw",
    "_sɛgɛn",
    "_sɛgɛsɛgɛ",
    "_sɛgɛsɛgɛli",
    "_sɛmɛntiya",
    "_sɛnɛ",
    "_sɛnɛfɛnw",
    "_sɛnɛkɛlaw",
    "_taamasiyɛn",
    "_taamasiyɛnw",
    "_taamasiɲɛw",
    "_taamasyɛn",
    "_tigɛ",
    "_tilancɛ",
    "_tiɲɛni",
    "_tiɲɛw",
    "_tɔgɔ",
    "_tɔgɔba",
    "_tɔnɔ",
    "_to",
    "_tɔw",
    "_tɔɔrɔ",
    "_tɛ",
    "_tɛmɛ",
    "_tɛmɛnen",
    "_tɛmɛsira",
    "_waleɲumandɔn",
    "_wɛrɛ",
    "_wɛrɛw",
    "_yɔrɔba",
    "_yɔrɔw",
    "_yɛrɛ",
    "_yɛrɛdɔn",
    "_yɛrɛmahɔrɔnya",
    "_yɛrɛmahɔrɔnyalen",
    "_ɔwɔ,",
    "_ɛndujamana",
    "_ɲagami",
    "_ɲagaminen",
    "_ɲi",
    "_ɲini",
    "_ɲinini",
    "_ɲininkali",
    "_ɲinw",
    "_ɲuman",
    "_ɲumanw",
    "_ɲɔgɔn",
    "_ɲɔgɔndan",
    "_ɲɔgɔndɛmɛ",
    "_ɲɔgɔnna",
    "_ɲɛ",
    "_ɲɛda",
    "_ɲɛfɔ",
    "_ɲɛfɔli",
    "_ɲɛgɛnw",
    "_ɲɛjirali",
    "_ɲɛmaa",
    "_ɲɛminɛ",
    "_ɲɛmɔgɔ",
    "_ɲɛmɔgɔw",
    "_ɲɛmɔgɔya",
    "_ɲɛnabɔ",
    "_ɲɛnabɔli",
    "_ɲɛnajɛ",
    "_ɲɛnajɛlamɔgɔw",
    "_ɲɛnama",
    "_ɲɛnamaya",
    "_ɲɛsin",
    "_ɲɛsinnen",
    "_ɲɛtaa",
]

BATCH_SIZE = 16
EPOCHS_PHASE1 = 10
EPOCHS_PHASE2 = 8
LR_PHASE1 = 2e-5
LR_PHASE2 = 1e-5
MIN_DELTA = 0.0001

# Montage Google Drive
print("🚀 Montage de Google Drive...")
drive.mount('/content/drive')

# ========== DATA AUGMENTATION ==========
def augment_triplet(triplet):
    """Simple data augmentation for text triplets"""
    augmented = triplet.copy()

    # Random word deletion (10% chance per word)
    for field in ['anchor', 'positive', 'negative']:
        words = triplet[field].split()
        if len(words) > 3 and random.random() < 0.3:
            del_idx = random.randint(0, len(words)-1)
            words.pop(del_idx)
            augmented[field] = ' '.join(words)

    # Synonym replacement (simplified example)
    replacements = {
         # Histoire/Culture
        'griot': ['jeli', 'djéli', 'àkòdànwò'],
        'histoire': ['tariku', 'lisite'],
        'empire': ['mansaya', 'sultani'],
        # Géographie
        'fleuve': ['ba', 'jiba'],
        'capitale': ['faaba', 'duguba'],
        'ville': ['dugu', 'sugo'],
        # Mathématiques
        'calculer': ['sanya', 'kalikuli'],
        'nombre': ['dɔgɔkun', 'nimɔrɔ'],
        'addition': ['fara', 'adision'],
        # Culture générale
        'cérémonie': ['seli', 'ɲɛnajɛ'],
        'tambour': ['dunun', 'tama']
    }

    for field in ['anchor', 'positive', 'negative']:
        for word, syns in replacements.items():
            if word in triplet[field]:
                augmented[field] = augmented[field].replace(word, random.choice(syns))

    return augmented

# ========== ENHANCED EVALUATOR ==========
def build_robust_evaluator(dataset, num_samples=200):
    """Build evaluator with proper train/test split"""
    # Sample evaluation data
    eval_samples = dataset.shuffle().select(range(min(num_samples, len(dataset))))

    queries = {}
    corpus = {}
    relevant_docs = {}

    for idx, sample in enumerate(eval_samples):
        queries[idx] = sample['anchor']
        corpus[2*idx] = sample['positive']
        corpus[2*idx+1] = sample['negative']
        relevant_docs[idx] = [2*idx]  # Only positive is relevant

    return evaluation.InformationRetrievalEvaluator(
        queries, corpus, relevant_docs,
        show_progress_bar=True,
        corpus_chunk_size=512,
        precision_recall_at_k=[3, 5, 10],
        name='bambara_eval'
    )

# ========== IMPROVED DATA LOADING ==========
def load_and_prepare_data():
    with zipfile.ZipFile(TRIPLETS_ZIP_PATH, 'r') as z:
        with z.open(z.namelist()[0]) as f:
            data = json.load(f)

    # Apply simple data augmentation
    augmented_data = []
    for item in data:
        augmented_data.append(item)
        # Add 1 augmented version for each original triplet (50% increase)
        if random.random() < 0.5:
            augmented_data.append(augment_triplet(item))

    # Split into phases with 80/20 train/val split
    random.shuffle(augmented_data)
    split_idx = int(0.8 * len(augmented_data))

    phase1_data, phase2_data = [], []
    for item in augmented_data[:split_idx]:  # Training data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            phase1_data.append(record)
        else:
            phase2_data.append(record)

    # Create validation sets
    val_phase1, val_phase2 = [], []
    for item in augmented_data[split_idx:]:  # Validation data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            val_phase1.append(record)
        else:
            val_phase2.append(record)

    return (
        Dataset.from_list(phase1_data),
        Dataset.from_list(val_phase1),
        Dataset.from_list(phase2_data),
        Dataset.from_list(val_phase2)
    )

# ========== ENHANCED TRAINING PHASE ==========
def train_phase(model, train_data, val_data, epochs, lr, phase_name):
    # Use multiple negative ranking loss for better convergence
    loss = losses.MultipleNegativesRankingLoss(model)

    # Create robust evaluator
    evaluator = build_robust_evaluator(val_data)

    # Enhanced training arguments
    training_args = SentenceTransformerTrainingArguments(
        output_dir=os.path.join(SAVE_DIR, phase_name),
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE*2,
        num_train_epochs=epochs,
        learning_rate=lr,
        weight_decay=0.01,
        warmup_ratio=0.1,
        eval_strategy="steps",
        logging_strategy="steps",
        eval_steps=100,
        save_steps=200,
        save_total_limit=2,
        load_best_model_at_end=True,
        metric_for_best_model="eval_bambara_eval_cosine_ndcg@10",
        greater_is_better=True,
        logging_dir=os.path.join(SAVE_DIR, "logs"),
        logging_steps=50,
        report_to=[],
        fp16=True
    )

    trainer = SentenceTransformerTrainer(
        model=model,
        args=training_args,
        train_dataset=train_data,
        eval_dataset=val_data,
        loss=loss,
        evaluator=evaluator,
        callbacks=[
            EarlyStoppingCallback(
                early_stopping_patience=3,
                early_stopping_threshold=MIN_DELTA
            )
        ],
    )

    print(f"\n⭐ Starting {phase_name} training with {len(train_data)} samples")
    trainer.train()

    # Final evaluation
    print("\n📊 Final evaluation results:")
    evaluator(model)

    model.save(os.path.join(SAVE_DIR, f"{phase_name}_final"))
    return model

# ========== MAIN ==========
def main():
    # Load model with better tokenizer handling
    model = SentenceTransformer(MODEL_NAME)

    # Ajout des tokens bambara
    tokenizer = model.tokenizer
    tokenizer.add_tokens(BAMBARA_TOKENS)
    model._first_module().auto_model.resize_token_embeddings(len(tokenizer))
    print(f"Tokenizer enrichi avec {len(BAMBARA_TOKENS)} tokens bambara")

    print("Model loaded with {} parameters".format(sum(p.numel() for p in model.parameters())))

    # Load and split data
    phase1_train, phase1_val, phase2_train, phase2_val = load_and_prepare_data()
    print(f"\nData loaded: {len(phase1_train)} phase1 train, {len(phase1_val)} phase1 val, "
          f"{len(phase2_train)} phase2 train, {len(phase2_val)} phase2 val samples")

    # Phase 1: Monolingual fine-tuning
    print("\n🔵 Phase 1: Monolingual Fine-tuning")
    model = train_phase(
        model, phase1_train, phase1_val,
        EPOCHS_PHASE1, LR_PHASE1, "phase1"
    )

    # Phase 2: Cross-lingual adaptation
    print("\n🟣 Phase 2: Cross-lingual Adaptation")
    model = train_phase(
        model, phase2_train, phase2_val,
        EPOCHS_PHASE2, LR_PHASE2, "phase2"
    )

    # Final save with better metadata
    model.save(os.path.join(SAVE_DIR, "final_model_bambara3"))
    print("\n✅ Training completed successfully!")

    # Test some examples
    test_queries = [
        ("Bonjour comment ça va?", "I ni sogoma, i ka kɛnɛ wa?"),
        ("Quelle est la capitale du Mali?", "Mali faaba ye mun ye?")
    ]
    for q1, q2 in test_queries:
        emb1 = model.encode(q1)
        emb2 = model.encode(q2)
        sim = util.cos_sim(emb1, emb2).item()
        print(f"Similarity between '{q1}' and '{q2}': {sim:.4f}")

if __name__ == "__main__":
    main()
```

```python
import os
import json
import zipfile
import random
from datasets import Dataset
from transformers import EarlyStoppingCallback
from sentence_transformers import SentenceTransformer, losses, util, evaluation
from sentence_transformers.trainer import SentenceTransformerTrainer
from sentence_transformers.training_args import SentenceTransformerTrainingArguments
from huggingface_hub import login, HfApi

# ========== CONFIG ==========
MODEL_NAME = "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
TRIPLETS_ZIP_PATH = "/content/drive/MyDrive/culture_donnees_sur_drive_sans_variance.zip"
HF_MODEL_NAME = "Fatoumataa/embedding_billingue_francais_bambara"  # Format: username/repo_name
LOCAL_SAVE_DIR = "/content/fine_tuned_model_temp"  # Dossier de sauvegarde temporaire

BATCH_SIZE = 16
EPOCHS_PHASE1 = 10
EPOCHS_PHASE2 = 8
LR_PHASE1 = 2e-5
LR_PHASE2 = 1e-5
MIN_DELTA = 0.0001




# ========== DATA AUGMENTATION ==========
def augment_triplet(triplet):
    """Simple data augmentation for text triplets"""
    augmented = triplet.copy()

    # Random word deletion (10% chance per word)
    for field in ['anchor', 'positive', 'negative']:
        words = triplet[field].split()
        if len(words) > 3 and random.random() < 0.3:
            del_idx = random.randint(0, len(words)-1)
            words.pop(del_idx)
            augmented[field] = ' '.join(words)

    # Synonym replacement (simplified example)
    replacements = {
         # Histoire/Culture
        'griot': ['jeli', 'djéli', 'àkòdànwò'],
        'histoire': ['tariku', 'lisite'],
        'empire': ['mansaya', 'sultani'],
        # Géographie
        'fleuve': ['ba', 'jiba'],
        'capitale': ['faaba', 'duguba'],
        'ville': ['dugu', 'sugo'],
        # Mathématiques
        'calculer': ['sanya', 'kalikuli'],
        'nombre': ['dɔgɔkun', 'nimɔrɔ'],
        'addition': ['fara', 'adision'],
        # Culture générale
        'cérémonie': ['seli', 'ɲɛnajɛ'],
        'tambour': ['dunun', 'tama']
    }

    for field in ['anchor', 'positive', 'negative']:
        for word, syns in replacements.items():
            if word in triplet[field]:
                augmented[field] = augmented[field].replace(word, random.choice(syns))

    return augmented

# ========== ENHANCED EVALUATOR ==========
def build_robust_evaluator(dataset, num_samples=200):
    """Build evaluator with proper train/test split"""
    # Sample evaluation data
    eval_samples = dataset.shuffle().select(range(min(num_samples, len(dataset))))

    queries = {}
    corpus = {}
    relevant_docs = {}

    for idx, sample in enumerate(eval_samples):
        queries[idx] = sample['anchor']
        corpus[2*idx] = sample['positive']
        corpus[2*idx+1] = sample['negative']
        relevant_docs[idx] = [2*idx]  # Only positive is relevant

    return evaluation.InformationRetrievalEvaluator(
        queries, corpus, relevant_docs,
        show_progress_bar=True,
        corpus_chunk_size=512,
        precision_recall_at_k=[3, 5, 10],
        name='bambara_eval'
    )

# ========== IMPROVED DATA LOADING ==========
def load_and_prepare_data():
    with zipfile.ZipFile(TRIPLETS_ZIP_PATH, 'r') as z:
        with z.open(z.namelist()[0]) as f:
            data = json.load(f)

    # Apply simple data augmentation
    augmented_data = []
    for item in data:
        augmented_data.append(item)
        # Add 1 augmented version for each original triplet (50% increase)
        if random.random() < 0.5:
            augmented_data.append(augment_triplet(item))

    # Split into phases with 80/20 train/val split
    random.shuffle(augmented_data)
    split_idx = int(0.8 * len(augmented_data))

    phase1_data, phase2_data = [], []
    for item in augmented_data[:split_idx]:  # Training data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            phase1_data.append(record)
        else:
            phase2_data.append(record)

    # Create validation sets
    val_phase1, val_phase2 = [], []
    for item in augmented_data[split_idx:]:  # Validation data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            val_phase1.append(record)
        else:
            val_phase2.append(record)

    return (
        Dataset.from_list(phase1_data),
        Dataset.from_list(val_phase1),
        Dataset.from_list(phase2_data),
        Dataset.from_list(val_phase2)
    )

# ========== ENHANCED TRAINING PHASE ==========
def train_phase(model, train_data, val_data, epochs, lr, phase_name):
    # Use multiple negative ranking loss for better convergence
    loss = losses.MultipleNegativesRankingLoss(model)

    # Create robust evaluator
    evaluator = build_robust_evaluator(val_data)

    # Enhanced training arguments
    training_args = SentenceTransformerTrainingArguments(
        output_dir=os.path.join(phase_name + "_output"),
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE*2,
        num_train_epochs=epochs,
        learning_rate=lr,
        weight_decay=0.01,  # Added regularization
        warmup_ratio=0.1,   # Better learning rate scheduling
        eval_strategy="steps",
        logging_strategy="steps",
        eval_steps=100,
        save_steps=200,
        save_total_limit=2,
        load_best_model_at_end=True,
        metric_for_best_model="eval_bambara_eval_cosine_ndcg@10",
        greater_is_better=True,
        logging_dir=os.path.join("logs"),
        logging_steps=50,
        report_to=[],
        fp16=True  # Enable mixed precision training
    )

    trainer = SentenceTransformerTrainer(
        model=model,
        args=training_args,
        train_dataset=train_data,
        eval_dataset=val_data,
        loss=loss,
        evaluator=evaluator,
        callbacks=[
            EarlyStoppingCallback(
                early_stopping_patience=3,
                early_stopping_threshold=MIN_DELTA
            )
        ],
    )

    print(f"\n⭐ Starting {phase_name} training with {len(train_data)} samples")
    trainer.train()

    # Final evaluation
    print("\n📊 Final evaluation results:")
    evaluator(model)

    return model





# ========== MAIN ==========
def main():
    # Load model with better tokenizer handling
    model = SentenceTransformer(MODEL_NAME)
    print("Model loaded with {} parameters".format(sum(p.numel() for p in model.parameters())))

    # Load and split data
    phase1_train, phase1_val, phase2_train, phase2_val = load_and_prepare_data()
    print(f"\nData loaded: {len(phase1_train)} phase1 train, {len(phase1_val)} phase1 val, "
          f"{len(phase2_train)} phase2 train, {len(phase2_val)} phase2 val samples")

    # Phase 1: Monolingual fine-tuning
    print("\n🔵 Phase 1: Monolingual Fine-tuning")
    model = train_phase(
        model, phase1_train, phase1_val,
        EPOCHS_PHASE1, LR_PHASE1, "phase1"
    )

    # Phase 2: Cross-lingual adaptation
    print("\n🟣 Phase 2: Cross-lingual Adaptation")
    model = train_phase(
        model, phase2_train, phase2_val,
        EPOCHS_PHASE2, LR_PHASE2, "phase2"
    )

    # ========== NOUVELLE PARTIE ENREGISTREMENT ==========
    print("\n💾 Début de l'enregistrement sur Hugging Face...")
    try:
        # Enregistrement sur Hugging Face Hub
        model.save_to_hub(
            repo_id=HF_MODEL_NAME,
            organization="Fatoumataa",
            commit_message="Modèle final français-bambara",
            private=True
        )
        print(f"\n✅ Modèle enregistré avec succès sur Hugging Face !")
        print(f"🔗 Accès: https://huggingface.co/{HF_MODEL_NAME}")

        # Sauvegarde locale de backup
        model.save(LOCAL_SAVE_DIR)
        print(f"📂 Backup local sauvegardé dans: {LOCAL_SAVE_DIR}")

    except Exception as e:
        print(f"\n❌ Erreur lors de l'enregistrement: {str(e)}")
        print("🔄 Tentative de sauvegarde locale complète...")
        model.save(LOCAL_SAVE_DIR + "_backup")
        print(f"⚠️ Modèle sauvegardé localement dans: {LOCAL_SAVE_DIR}_backup")
        print("Vous pouvez uploader manuellement ce dossier sur Hugging Face Hub")

    # Tests finaux
    test_queries = [
        ("Bonjour comment ça va?", "I ni sogoma, i ka kɛnɛ wa?"),
        ("Quelle est la capitale du Mali?", "Mali faaba ye mun ye?")
    ]
    for q1, q2 in test_queries:
        emb1 = model.encode(q1)
        emb2 = model.encode(q2)
        sim = util.cos_sim(emb1, emb2).item()
        print(f"Similarity between '{q1}' and '{q2}': {sim:.4f}")

if __name__ == "__main__":
    main()
```

```python
# 
import os
import json
import zipfile
import random
from datasets import Dataset
from transformers import EarlyStoppingCallback
from sentence_transformers import SentenceTransformer, losses, util, evaluation
from sentence_transformers.trainer import SentenceTransformerTrainer
from sentence_transformers.training_args import SentenceTransformerTrainingArguments
from huggingface_hub import login, HfApi
# from google.colab import drive


 

# Charger votre modèle
MODEL_NAME = "Fatoumataa/embedding_billingue_francais_bambara"
TRIPLETS_ZIP_PATH = "/content/drive/MyDrive/fichier2_sans_doublons.zip"
HF_MODEL_NAME = "Fatoumataa/embedding_billingue_francais_bambara_version3"  
LOCAL_SAVE_DIR = "/content/fine_tuned_model_temp"  # Dossier de sauvegarde temporaire

BATCH_SIZE = 16
EPOCHS_PHASE1 = 6
EPOCHS_PHASE2 = 6
LR_PHASE1 = 2e-5
LR_PHASE2 = 1e-5
MIN_DELTA = 0.0001

# ========== DATA AUGMENTATION ==========
def augment_triplet(triplet):
    """Simple data augmentation for text triplets"""
    augmented = triplet.copy()

    # Random word deletion (10% chance per word)
    for field in ['anchor', 'positive', 'negative']:
        words = triplet[field].split()
        if len(words) > 3 and random.random() < 0.3:
            del_idx = random.randint(0, len(words)-1)
            words.pop(del_idx)
            augmented[field] = ' '.join(words)

    # Synonym replacement (simplified example)
    replacements = {
         # Histoire/Culture
        'griot': ['jeli', 'djéli', 'àkòdànwò'],
        'histoire': ['tariku', 'lisite'],
        'empire': ['mansaya', 'sultani'],
        # Géographie
        'fleuve': ['ba', 'jiba'],
        'capitale': ['faaba', 'duguba'],
        'ville': ['dugu', 'sugo'],
        # Mathématiques
        'calculer': ['sanya', 'kalikuli'],
        'nombre': ['dɔgɔkun', 'nimɔrɔ'],
        'addition': ['fara', 'adision'],
        # Culture générale
        'cérémonie': ['seli', 'ɲɛnajɛ'],
        'tambour': ['dunun', 'tama']
    }

    for field in ['anchor', 'positive', 'negative']:
        for word, syns in replacements.items():
            if word in triplet[field]:
                augmented[field] = augmented[field].replace(word, random.choice(syns))

    return augmented

# ========== ENHANCED EVALUATOR ==========
def build_robust_evaluator(dataset, num_samples=200):
    """Build evaluator with proper train/test split"""
    # Sample evaluation data
    eval_samples = dataset.shuffle().select(range(min(num_samples, len(dataset))))

    queries = {}
    corpus = {}
    relevant_docs = {}

    for idx, sample in enumerate(eval_samples):
        queries[idx] = sample['anchor']
        corpus[2*idx] = sample['positive']
        corpus[2*idx+1] = sample['negative']
        relevant_docs[idx] = [2*idx]  # Only positive is relevant

    return evaluation.InformationRetrievalEvaluator(
        queries, corpus, relevant_docs,
        show_progress_bar=True,
        corpus_chunk_size=512,
        precision_recall_at_k=[3, 5, 10],
        name='bambara_eval'
    )

# ========== IMPROVED DATA LOADING ==========
def load_and_prepare_data():
    with zipfile.ZipFile(TRIPLETS_ZIP_PATH, 'r') as z:
        with z.open(z.namelist()[0]) as f:
            data = json.load(f)

    # Apply simple data augmentation
    augmented_data = []
    for item in data:
        augmented_data.append(item)
        # Add 1 augmented version for each original triplet (50% increase)
        if random.random() < 0.5:
            augmented_data.append(augment_triplet(item))

    # Split into phases with 80/20 train/val split
    random.shuffle(augmented_data)
    split_idx = int(0.8 * len(augmented_data))

    phase1_data, phase2_data = [], []
    for item in augmented_data[:split_idx]:  # Training data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            phase1_data.append(record)
        else:
            phase2_data.append(record)

    # Create validation sets
    val_phase1, val_phase2 = [], []
    for item in augmented_data[split_idx:]:  # Validation data
        record = {
            "anchor": item["anchor"],
            "positive": item["positive"],
            "negative": item["negative"],
            "label": 1.0
        }
        if item["langue"] == item.get("positive_langue", item["langue"]):
            val_phase1.append(record)
        else:
            val_phase2.append(record)

    return (
        Dataset.from_list(phase1_data),
        Dataset.from_list(val_phase1),
        Dataset.from_list(phase2_data),
        Dataset.from_list(val_phase2)
    )

# ========== ENHANCED TRAINING PHASE ==========
def train_phase(model, train_data, val_data, epochs, lr, phase_name):
    # Use multiple negative ranking loss for better convergence
    loss = losses.MultipleNegativesRankingLoss(model)

    # Create robust evaluator
    evaluator = build_robust_evaluator(val_data)

    # Enhanced training arguments
    training_args = SentenceTransformerTrainingArguments(
        output_dir=os.path.join(phase_name + "_output"),
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE*2,
        num_train_epochs=epochs,
        learning_rate=lr,
        weight_decay=0.01,  # Added regularization
        warmup_ratio=0.1,   # Better learning rate scheduling
        eval_strategy="steps",
        logging_strategy="steps",
        eval_steps=100,
        save_steps=200,
        save_total_limit=2,
        load_best_model_at_end=True,
        metric_for_best_model="eval_bambara_eval_cosine_ndcg@10",
        greater_is_better=True,
        logging_dir=os.path.join("logs"),
        logging_steps=50,
        report_to=[],
        fp16=True  # Enable mixed precision training
    )

    trainer = SentenceTransformerTrainer(
        model=model,
        args=training_args,
        train_dataset=train_data,
        eval_dataset=val_data,
        loss=loss,
        evaluator=evaluator,
        callbacks=[
            EarlyStoppingCallback(
                early_stopping_patience=3,
                early_stopping_threshold=MIN_DELTA
            )
        ],
    )

    print(f"\n⭐ Starting {phase_name} training with {len(train_data)} samples")
    trainer.train()

    # Final evaluation
    print("\n📊 Final evaluation results:")
    evaluator(model)

    return model





# ========== MAIN ==========
def main():
    # Load model with better tokenizer handling
    model = SentenceTransformer(MODEL_NAME)
    print("Model loaded with {} parameters".format(sum(p.numel() for p in model.parameters())))

    # Load and split data
    phase1_train, phase1_val, phase2_train, phase2_val = load_and_prepare_data()
    print(f"\nData loaded: {len(phase1_train)} phase1 train, {len(phase1_val)} phase1 val, "
          f"{len(phase2_train)} phase2 train, {len(phase2_val)} phase2 val samples")

    # Phase 1: Monolingual fine-tuning
    print("\n🔵 Phase 1: Monolingual Fine-tuning")
    model = train_phase(
        model, phase1_train, phase1_val,
        EPOCHS_PHASE1, LR_PHASE1, "phase1"
    )

    # Phase 2: Cross-lingual adaptation
    print("\n🟣 Phase 2: Cross-lingual Adaptation")
    model = train_phase(
        model, phase2_train, phase2_val,
        EPOCHS_PHASE2, LR_PHASE2, "phase2"
    )

    # ========== NOUVELLE PARTIE ENREGISTREMENT ==========
    print("\n💾 Début de l'enregistrement sur Hugging Face...")
    try:
        # Enregistrement sur Hugging Face Hub
        model.save_to_hub(
            repo_id=HF_MODEL_NAME,
            organization="Fatoumataa",
            commit_message="Modèle final français-bambara",
            private=True
        )
        print(f"\n✅ Modèle enregistré avec succès sur Hugging Face !")
        print(f"🔗 Accès: https://huggingface.co/{HF_MODEL_NAME}")

        # Sauvegarde locale de backup
        model.save(LOCAL_SAVE_DIR)
        print(f"📂 Backup local sauvegardé dans: {LOCAL_SAVE_DIR}")

    except Exception as e:
        print(f"\n❌ Erreur lors de l'enregistrement: {str(e)}")
        print("🔄 Tentative de sauvegarde locale complète...")
        model.save(LOCAL_SAVE_DIR + "_backup")
        print(f"⚠️ Modèle sauvegardé localement dans: {LOCAL_SAVE_DIR}_backup")
        print("Vous pouvez uploader manuellement ce dossier sur Hugging Face Hub")

    # Tests finaux
    test_queries = [
        ("Bonjour comment ça va?", "I ni sogoma, i ka kɛnɛ wa?"),
        ("Quelle est la capitale du Mali?", "Mali faaba ye mun ye?")
    ]
    for q1, q2 in test_queries:
        emb1 = model.encode(q1)
        emb2 = model.encode(q2)
        sim = util.cos_sim(emb1, emb2).item()
        print(f"Similarity between '{q1}' and '{q2}': {sim:.4f}")

if __name__ == "__main__":
    main()
```
